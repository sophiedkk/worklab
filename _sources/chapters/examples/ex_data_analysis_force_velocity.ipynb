# Analysis of Force Velocity profiling

<b>Protocol ergometer:</b>
Repeated individualised sprint protocol at varied resistances

<b>Description of the sprint test: </b>
First 10s sprint will be at resistance 0.012, other sprints will be individualised based on outcomes of this first sprint

<b>Outcome: </b>
Force-velocity curve

First, let us import the right package(s)
As a filename you can select a folder with all participant files, or a specific folder for an athlete

```python
import worklab as wl
import os

filename = os.getcwd()
filename_fv = os.path.join('r',filename, 'example_data', 'Force velocity example')
```

Let's define some user-specific variables

```python
athlete = 'PP01' # Change athlete, make sure this is equal to your folder name
classification = 'LP'
sport = 'WR'
```

Make sure that all sprints are organised in a separate Excel file! Listed as sprints '1', '2', '3' etc.

If an additional sprint was completed at a resistance 0.008, this can be added as a 0.xlsx file.

```python
var_fvc, outcomes_mean_all, data_pbp_all = wl.ana.force_vel_profiling(filename_fv, athlete='PP01', classification='LP', n_sprints=7, var='mean', unit='kmh')
```

Force-velocity profiling is succesfull, you should have multiple excel files and figures in your outcomes folder.
